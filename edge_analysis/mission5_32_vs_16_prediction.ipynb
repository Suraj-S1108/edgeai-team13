# EDGE AI Team 13 - Mission 5: 32-32 vs 16-16 Model Comparison

Detailed side-by-side analysis of individual predictions, residuals, and error distributions between the deployed 32-32 model and legacy 16-16 model.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_395 = pd.read_csv('edge_analysis/mission5_results/mission5_all_395_validation_comparison.csv')
print(f'Total validation comparisons: {len(df_395)}')
print(df_395.head())
```

## 1. Summary Statistics of Predictions and Errors


```python
print('16-16 Error Mean:   ', df_395['Error_16_16'].mean())
print('16-16 Error Median: ', df_395['Error_16_16'].median())
print('16-16 Error P95:    ', np.percentile(df_395['Error_16_16'], 95))
print('---')
print('32-32 Error Mean:   ', df_395['Error_32_32'].mean())
print('32-32 Error Median: ', df_395['Error_32_32'].median())
print('32-32 Error P95:    ', np.percentile(df_395['Error_32_32'], 95))
```

## 2. Head-to-Head Win Rate

Comparing which model makes a more accurate prediction on each sample.


```python
win_32 = (df_395['Error_32_32'] < df_395['Error_16_16']).sum()
win_16 = (df_395['Error_16_16'] < df_395['Error_32_32']).sum()
print(f'32-32 was more accurate on {win_32} out of {len(df_395)} samples ({win_32/len(df_395)*100:.1f}%)')
print(f'16-16 was more accurate on {win_16} out of {len(df_395)} samples ({win_16/len(df_395)*100:.1f}%)')
```
