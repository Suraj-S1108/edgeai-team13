# EDGE AI Team 13 - Mission 5: Model Analysis and Pareto Comparison

Comprehensive empirical analysis of original versus cleaned datasets, network depth, width scaling, and edge deployment trade-offs.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_comp = pd.read_csv('edge_analysis/mission5_results/mission5_complete_model_comparison.csv')
df_train_val = pd.read_csv('edge_analysis/mission5_results/mission5_train_validation_complete.csv')
df_cleaning = pd.read_csv('edge_analysis/mission5_results/cleaning_audit.csv')
print('Loaded Mission 5 experiment records successfully.')
```

## 1. Cleaning Audit Summary

Quantifying the impact of glitch removal on the training dataset.


```python
print(df_cleaning.to_string(index=False))
```

## 2. Complete Model Comparison

Comparing MAE, RMSE, R2, and parameter counts across all models.


```python
cleaned_models = df_comp[df_comp['Dataset'] == 'Cleaned'].sort_values('MAE')
print(cleaned_models[['Model', 'Parameters', 'MAE', 'RMSE', 'R2']].to_string(index=False))
```

## 3. Depth and Width Analysis

Evaluating the performance gains from adding depth vs increasing width.


```python
two_layers = df_train_val[(df_train_val['Dataset'] == 'Cleaned') & (df_train_val['Hidden Layers'] == 2)]
print('Two Hidden Layer Performance Scaling:')
print(two_layers[['Model', 'Parameters', 'Train MAE', 'Validation MAE', 'MAE Gap']].to_string(index=False))
```

## 4. Edge AI Pareto-Optimal Model Selection

Why NN_2L_32_32 is selected:
- 24.2% error reduction over 16-16 (5.22 days vs 6.88 days)
- Low memory requirement (1,281 parameters = 5.12 KiB FP32)
- Avoids severe diminishing returns of 64-64 (4,609 parameters for only 0.09 days MAE gain)


```python
best_model = cleaned_models[cleaned_models['Model'] == 'NN_2L_32_32'].iloc[0]
print(f'Selected Model: {best_model["Model"]}')
print(f'Parameters:     {best_model["Parameters"]}')
print(f'Validation MAE: {best_model["MAE"]:.4f} days')
print(f'Validation R2:  {best_model["R2"]:.4f}')
```
